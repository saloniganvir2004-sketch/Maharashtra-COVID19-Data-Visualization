```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df= pd.read_csv('/Maharashtra Latest Covid Cases.csv')
df.head(5)
x=df['Deceased']
y=df['Active Cases']
z=df['Fatality Rate(%)']
a=df['Recovery Rate(%)']
b=df['Districts']
plt.plot(x,y)
plt.title('Line Graph')
plt.xlabel('Deceased')
plt.ylabel('Active Cases')
plt.show()
plt.xlabel('Deceased',fontsize=12)
plt.ylabel('Active Cases',fontsize=12)
plt.bar(b,x)
plt.title('Bar Graph')
plt.show()
plt.scatter(x,y)
plt.title('Scatter Graph')
plt.xlabel('Deceased')
plt.ylabel('Active Cases')
plt.show()
plt.pie(y,labels=y)
plt.title('Pie Chart')
plt.show()
plt.hist(y,bins=10)
plt.title('Histograph')
plt.xlabel('Deceased')
plt.ylabel('Active Cases')
plt.show()
plt.boxplot(x)
plt.title('Boxplot')
plt.xlabel('Districts')
plt.ylabel('Active Cases')
plt.show()
```
